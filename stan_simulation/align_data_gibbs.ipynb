```python
import os
import shutil
import re

# Base directories
input_base_dir = 'ten_V200K10/nofix/gibbs_raw'
output_base_dir = 'ten_V200K10/nofix/gibbs'
# List all directories under input_base_dir starting with 'largesim-V200-K10-N'
size_dirs = [
    d for d in os.listdir(input_base_dir)
    if os.path.isdir(os.path.join(input_base_dir, d)) and d.startswith('largesim-V200-K10-N')
]

# Loop over each size directory
for size_dir in size_dirs:
    # Extract the 'size' from the directory name
    match_size = re.search(r'N(\d+)', size_dir)
    if match_size:
        size = match_size.group(1)
    else:
        continue  # Skip if size not found

    size_dir_path = os.path.join(input_base_dir, size_dir)

    # List all files in the size directory
    files = os.listdir(size_dir_path)
    for file_name in files:
        # Match filenames of the form gibbs-samples-N-{size}-D-10-largesim{i}.csv
        pattern = rf'gibbs-samples-N-{size}-D-10-largesim(\d+)\.csv'
        match_file = re.match(pattern, file_name)
        if match_file:
            i = match_file.group(1)
            # Create the output directory if it doesn't exist
            output_dir = os.path.join(output_base_dir, i)
            os.makedirs(output_dir, exist_ok=True)
            # Define source and destination file paths
            src_file = os.path.join(size_dir_path, file_name)
            dst_file = os.path.join(output_dir, f'gibbs-samples-N-{size}.csv')
            # Move the file to the new location with the new name
            shutil.move(src_file, dst_file)
        else:
            continue  # Skip files that don't match the pattern
```
