```python

def sgns_likelihood(target_embedding, context_embedding, i, j, x=None):
    rhos = target_embedding[i]      
    alphas = context_embedding[j]   

    logits = np.sum(rhos * alphas, axis=-1)
    if x is not None:
        x = 2 * x - 1
        logits *= x

    ps = 1.0 / (1.0 + np.exp(-logits))
    log_ps = np.log(ps)
    return log_ps


# -- data --
data_path = os.path.join("../../data/movielens-test.json")
with open(data_path) as f:
    data = json.load(f)
        
test_data = data['data']
vocab = data['vocabulary']

# -- samples --
sample_folder = '../../results-movielens-train/nofix/vi/movielens-train'
size = 850000
samples = load_fit_by_size(size, sample_folder)

word_vectors, context_vectors = extract_word_and_context_vectors_vi(samples.variational_sample_pd, vocab, D=20)


# --- likelihood ---
mean_target_vec = np.mean(word_vectors, axis=2)
mean_context_vec = np.mean(context_vectors, axis=2)

target_word_indices = []
context_word_indices = []
xx = []
for entry in test_data:
    target_word_indices.append(vocab[entry['v']])
    context_word_indices.append(vocab[entry['w']])
    xx.append(entry['x'])


log_probs = sgns_likelihood(
    mean_target_vec, 
    mean_context_vec, 
    np.array(target_word_indices), 
    np.array(context_word_indices),
    x=np.array(xx)
)

total_log_likelihood = log_probs.mean()
total_log_likelihood
```
