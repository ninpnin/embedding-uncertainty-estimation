```python
import os
import pickle
import json
from parsing_library import *
```

This script converts the full hmc stan pickles to ones only containing samples - greatly reducing disk space.


observe that it overrides the old files.

```python
fit_base_dir = "../../results_ten_V200_K10/nofix/hmc"

for directory in os.listdir(fit_base_dir):
    files = os.listdir(os.path.join(fit_base_dir, directory))
    for f in files:
        if f[-3:] == 'pkl':
            file_path = os.path.join(fit_base_dir,directory,f)
            with open(file_path, 'rb') as fn:
                fit = pickle.load( fn )
            fit = {'word_vectors':fit['word_vectors'], 'context_vectors':fit['context_vectors']} 
            with open(file_path, 'wb') as f:
                pickle.dump(fit, f)
            

```
